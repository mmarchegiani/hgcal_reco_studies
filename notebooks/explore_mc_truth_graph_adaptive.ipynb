## Ambiguous tracksters: two unrelated truths (x-z view)

Event 0, CLUE3DHigh tracksters matched to two **unrelated** TruthFG particles: the best-scoring truth, plus the best-scoring truth that is not its ancestor or descendant and does not share any ancestor with it (ancestor chain via `parentGraphId` → `graphId`, plus a different `genRootGraphId`, because some parents are missing from the collection).

The top-2 links are almost always the same particle at two graph levels (e.g. π0 → γ). Unrelated truths split the trackster energy, so the second score can't go below ~0.5 (no such pair in event 0 has 2nd score < 0.2). The cut is therefore `max_second_score`, and examples are ranked by the second truth's score (most ambiguous first).

Trackster: 1σ ellipse centred on the barycenter, built from `eVector0` and `EV1..3` and projected onto x-z (only `eVector0` is stored, so the transverse spread is taken as isotropic with variance (EV2+EV3)/2).
Truths: arrows from the boundary crossing along the boundary momentum. Truths without a boundary have no stored vertex, so they are drawn dotted from the origin along their (eta, phi).

```python
import numpy as np
from matplotlib.patches import Ellipse, FancyArrowPatch

ievt = 0
max_second_score = 0.6  # unrelated truths share the energy, so the 2nd score is >~0.5
evt = df[ievt]
ts = evt.ticlTrackstersCLUE3DHigh
truth = evt.TruthFG
assoc = evt.CLUE3DHighToTruthFG
nlinks = assoc.CLUE3DHighToTruthFG_nCLUE3DHighToTruthFGLinks
ev_links = ak.unflatten(assoc.CLUE3DHighToTruthFGLinks_index, nlinks)
ev_scores = ak.unflatten(assoc.CLUE3DHighToTruthFGLinks_score, nlinks)  # sorted ascending: lower = better

# ancestry: parentGraphId refers to graphId (not to the array index); some parents are not stored
graph_id = ak.to_numpy(truth.TruthFG_graphId)
parent_id = ak.to_numpy(truth.TruthFG_parentGraphId)
root_id = ak.to_numpy(truth.TruthFG_genRootGraphId)
gid_to_idx = {g: k for k, g in enumerate(graph_id)}

def lineage(j):
    """graphIds of truth j and all its stored (or first missing) ancestors."""
    out, p = {graph_id[j]}, parent_id[j]
    while p >= 0:
        out.add(p)
        if p not in gid_to_idx:
            break
        p = parent_id[gid_to_idx[p]]
    return out

def unrelated(a, b):
    return root_id[a] != root_id[b] and not (lineage(a) & lineage(b))

# for each trackster: best truth + best-scoring truth unrelated to it
pairs = []  # (trackster, (truth_a, score_a), (truth_b, score_b))
for i, (lks, scs) in enumerate(zip(ev_links.tolist(), ev_scores.tolist())):
    if len(lks) < 2:
        continue
    for b, sb in zip(lks[1:], scs[1:]):
        if unrelated(lks[0], b):
            if sb < max_second_score:
                pairs.append((i, (lks[0], scs[0]), (b, sb)))
            break
pairs.sort(key=lambda p: p[2][1])
print(f"{len(pairs)} / {len(ts.ticlTrackstersCLUE3DHigh_raw_energy)} tracksters with an unrelated 2nd truth, score < {max_second_score}")

def ts_get(var, i):
    return float(ts[f"ticlTrackstersCLUE3DHigh_{var}"][i])

def truth_start_dir(j):
    """Start point and unit direction of truth j. Uses the boundary crossing if present,
    otherwise the nominal origin (no production vertex is stored) with eta/phi direction."""
    if truth.TruthFG_hasBoundary[j]:
        start = np.array([truth[f"TruthFG_boundary{a}"][j] for a in "XYZ"], dtype=float)
        d = np.array([truth[f"TruthFG_boundaryP{a}"][j] for a in "xyz"], dtype=float)
    else:
        start = np.zeros(3)
        eta, phi = float(truth.TruthFG_eta[j]), float(truth.TruthFG_phi[j])
        d = np.array([np.cos(phi), np.sin(phi), np.sinh(eta)])
    return start, d / np.linalg.norm(d)

def truth_eta_phi(j):
    """Direction at the boundary crossing if present, otherwise at production."""
    if truth.TruthFG_hasBoundary[j]:
        return float(truth.TruthFG_boundaryEta[j]), float(truth.TruthFG_boundaryPhi[j])
    return float(truth.TruthFG_eta[j]), float(truth.TruthFG_phi[j])

def pair_dr_efrac(a, b):
    """deltaR between truths a and b, and E_a / (E_a + E_b)."""
    (eta_a, phi_a), (eta_b, phi_b) = truth_eta_phi(a), truth_eta_phi(b)
    dphi = (phi_a - phi_b + np.pi) % (2 * np.pi) - np.pi
    e_a, e_b = float(truth.TruthFG_energy[a]), float(truth.TruthFG_energy[b])
    return np.hypot(eta_a - eta_b, dphi), e_a / (e_a + e_b)

def xz_ellipse(i, nsigma=1.0):
    """Project the trackster PCA onto the x-z plane. Only eVector0 is stored, so the
    transverse spread is approximated as isotropic with variance (EV2+EV3)/2."""
    v = np.array([ts_get(f"eVector0_{a}", i) for a in "xyz"])
    ev1, ev2, ev3 = (ts_get(f"EV{k}", i) for k in (1, 2, 3))
    cov = ev1 * np.outer(v, v) + 0.5 * (ev2 + ev3) * (np.eye(3) - np.outer(v, v))
    cov_zx = cov[np.ix_([2, 0], [2, 0])]  # plot axes: (z, x)
    lam, vec = np.linalg.eigh(cov_zx)
    angle = np.degrees(np.arctan2(vec[1, 1], vec[0, 1]))
    w, h = 2 * nsigma * np.sqrt(np.clip(lam[::-1], 0, None))
    return w, h, angle, v

def plot_trackster(ax, i, matched):
    """matched: list of (truth index, score) to draw as arrows."""
    bz, bx = ts_get("barycenter_z", i), ts_get("barycenter_x", i)
    w, h, angle, v = xz_ellipse(i)
    ax.add_patch(Ellipse((bz, bx), w, h, angle=angle, fc="tab:blue", ec="tab:blue", alpha=0.35, lw=1.5,
                         label=f"trackster {i} ({ts_get('raw_energy', i):.1f} GeV)"))
    ax.plot(bz, bx, "x", color="tab:blue")
    # principal axis (eVector0) projected on x-z, half-length sqrt(EV1)
    l1 = np.sqrt(ts_get("EV1", i))
    ax.plot([bz - l1 * v[2], bz + l1 * v[2]], [bx - l1 * v[0], bx + l1 * v[0]], "--", color="tab:blue", lw=1)

    zs, xs = [bz - w, bz + w], [bx - w, bx + w]
    for k, ((j, score), col) in enumerate(zip(matched, ["tab:red", "tab:green"])):
        start, d = truth_start_dir(j)
        # arrow runs from start to a bit past the trackster barycenter in z
        z_end = bz + np.sign(d[2]) * max(10.0, l1)
        t = (z_end - start[2]) / d[2] if abs(d[2]) > 1e-3 else 20.0
        end = start + t * d
        ls = "-" if truth.TruthFG_hasBoundary[j] else ":"
        pu = ", PU" if truth.TruthFG_isPileup[j] else ""
        ax.add_patch(FancyArrowPatch((start[2], start[0]), (end[2], end[0]), arrowstyle="-|>", mutation_scale=15,
                                     color=col, lw=2, ls=ls, clip_on=True,
                                     label=f"#{k+1} pdg {truth.TruthFG_pdgId[j]}, {truth.TruthFG_energy[j]:.1f} GeV{pu}, "
                                           f"score {score:.3f}"))
        if truth.TruthFG_hasBoundary[j]:
            ax.plot(start[2], start[0], "o", color=col, ms=5)
            zs.append(start[2]); xs.append(start[0])
        zs.append(end[2]); xs.append(end[0])
    pad = 5
    ax.set_xlim(min(zs) - pad, max(zs) + pad)
    ax.set_ylim(min(xs) - pad, max(xs) + pad)
    ax.set_xlabel("z [cm]")
    ax.set_ylabel("x [cm]")
    if len(matched) == 2:
        dr, efrac = pair_dr_efrac(matched[0][0], matched[1][0])
        ax.text(0.03, 0.04, f"$\\Delta R_{{12}}$ = {dr:.3f}\n$E_1/(E_1+E_2)$ = {efrac:.2f}", transform=ax.transAxes,
                fontsize=13, va="bottom", ha="left", bbox=dict(fc="white", ec="0.6", alpha=0.9))
    ax.legend(fontsize=10, loc="best")

# show the most ambiguous ones (lowest score of the unrelated 2nd truth)
nshow = 6
fig, axes = plt.subplots(2, 3, figsize=(24, 15))
for ax, (i, ta, tb) in zip(axes.flat, pairs[:nshow]):
    plot_trackster(ax, i, [ta, tb])
for ax in axes.flat[len(pairs[:nshow]):]:
    ax.set_visible(False)
fig.suptitle(f"Event {ievt}: CLUE3DHigh tracksters matched to two unrelated TruthFG particles (2nd score < {max_second_score})\n"
             "solid arrow: from boundary crossing; dotted: from origin (no boundary)", fontsize=16)
fig.tight_layout()
fig.savefig("plots/clue3d_truthfg_ambiguous_xz.png", dpi=100)
```
